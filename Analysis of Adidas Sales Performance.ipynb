Total Sales Over Time (Trend Line)

```python
import pandas as pd
import matplotlib.pyplot as plt

df['Invoice Date'] = pd.to_datetime(df1['Invoice Date'])
sales_trend = df.groupby('Invoice Date')['Total Sales'].sum()
plt.figure(figsize=(5,4))  # smaller graph
plt.plot(sales_trend)
plt.title("Total Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

Sales by Region (Bar Chart)

```python
region_sales = df.groupby('Region')['Total Sales'].sum()
plt.figure(figsize=(3,2))  # smaller graph

region_sales.plot(kind='bar')
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

Sales by Product Category (Bar Chart)

```python
product_sales = df.groupby('Product')['Total Sales'].sum()
plt.figure(figsize=(4,2))
product_sales.plot(kind='bar')
plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

Profit vs Sales (Scatter Plot)

```python
plt.figure(figsize=(5,3))
plt.scatter(df['Total Sales'], df['Operating Profit'])
plt.title("Sales vs Profit")
plt.xlabel("Total Sales")
plt.ylabel("Operating Profit")
plt.show()
```

Sales Method Distribution (Pie Chart)

```python
method_sales = df.groupby('Sales Method')['Total Sales'].sum()

plt.figure()
method_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title("Sales Method Contribution")
plt.ylabel("")
plt.show()
```

```python
Monthly Sales Trend
```

```python
df['Month'] = df['Invoice Date'].dt.to_period('M')
monthly_sales = df.groupby('Month')['Total Sales'].sum()

plt.figure(figsize=(4,3))
monthly_sales.plot()
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

Operating Margin by Product (Bar Chart)

```python
margin = df.groupby('Product')['Operating Margin'].mean()

plt.figure(figsize=(4,3))
margin.plot(kind='bar')
plt.title("Average Operating Margin by Product")
plt.xlabel("Product")
plt.ylabel("Margin")
plt.xticks(rotation=45)
plt.show()
```

Units Sold vs Price (Scatter Plot)

```python
plt.figure(figsize=(4,3))
plt.scatter(df['Price per Unit'], df['Units Sold'])
plt.title("Price vs Units Sold")
plt.xlabel("Price per Unit")
plt.ylabel("Units Sold")
plt.show()
```

Sales by State

```python
state = df.groupby('State')['Total Sales'].sum().nlargest(10)

plt.figure(figsize=(4,4))
state.plot(kind='bar')

plt.title("Top 10 States by Sales")
plt.xlabel("State")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Sales by City (Top 10)

```python
city = df.groupby('City')['Total Sales'].sum().nlargest(10)
plt.figure(figsize=(4,3))
city.plot(kind='bar')
plt.title("Top 10 Cities by Sales")
plt.show()
```

Units Sold by Product

```python
units_prod = df.groupby('Product')['Units Sold'].sum()
plt.figure(figsize=(4,3))
units_prod.plot(kind='bar')
plt.title("Units Sold by Product")
plt.show()
```
